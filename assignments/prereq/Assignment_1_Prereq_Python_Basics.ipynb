```python
# Question 1: Given a list of tuples, find the number of occurrences of each tuple in the list.


def get_occ_tuple(lis):
  d = {}
  for item in lis:
      if not d.get(item):
          d[item] = 0
      d[item] += 1
  return d


lis = [(1, 2, 3, 4, 5), (1, 2), (1, 2), (1, 2), (1, 2), (2, 3, 4), (3, 4)]
print(get_occ_tuple(lis))
```

```python
# Question 2: Create a list of 10 random numbers between 1-100. Create a balanced BST (Binary search tree) from this array

import random

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
class BST:
    def __init__(self, array: list) -> None:
        self.array = array
        self.n = len(self.array)
        self.root = None
        
    def sort(self) -> None:
        # Bubble Sort
        n = len(self.array)
    
        for i in range(n):
            for j in range(n-i-1):
                if self.array[j] > self.array[j+1]:
                    self.array[j], self.array[j+1] = self.array[j+1], self.array[j]
    
    def build_bst(self, array: list) -> Node:
        # Build the Balanced BST
      
        if not array:
            return None

        mid = (len(array))//2
        root = Node(array[mid])
        root.left = self.build_bst(array[:mid])
        root.right = self.build_bst(array[mid+1:])
        
        return root
    
    def print_preorder_bst(self, node: Node) -> None:
        # Pre-Order
        if not node:
            return
        
        print(node.value, end='|') 
        self.print_preorder_bst(node.left)
        self.print_preorder_bst(node.right)
        
        return
    
    def direction(self) -> None:
        self.sort()
        self.root = self.build_bst(self.array)
        self.print_preorder_bst(self.root)


# Random List Generation
lis = random.sample(range(1, 100), 10)
print(f"Randomly generated list : {lis}")

obj = BST(lis)
obj.direction()

# Printing in 2D
space_count = 8
def print_in_2d(root: Node, space: int):
    if root == None:
        return
    space += space_count
    
    print_in_2d(root.right, space)
    print()
    print(" "*(space - space_count), end="") # 
    print(root.value)  
    print_in_2d(root.left, space)

print_in_2d(obj.root, 0) 
```

```python
# Question 3: Given the following dictionary:Create a list of all the keys of the dictionary, which can be formed from the values present in the dictionary.


ldict = {
    'black': 'r',
    'hero': 'e',
    'go': 'g',
    'clue': 'i',
    'mean': 'q',
    'groan': 'o',
    'sin': 'p',
    'pint': 'u',
    'tone': 'n',
    'graze': 's',
    'sea': 't',
    'plant': 'a'
}


new_dict = {}
words_cant_be_made = []
words_can_be_made = set(ldict.keys())

for word in ldict.keys():
    for letter in word:
        if not new_dict.get(letter):
            new_dict[letter] = []
        new_dict[letter].append(word)
print("\nFirst Step")
print(new_dict)
for item in ldict.values():
    if new_dict.get(item):
        new_dict.pop(item)
print("\nSecond Step")
print(new_dict)
        
words_cant_be_made = []
for val in new_dict.values():
    words_cant_be_made.extend(val)
words_cant_be_made = set(words_cant_be_made)
words_can_be_made = list(words_can_be_made - words_cant_be_made)
print(f"\nWords that can be made : {words_can_be_made}")
```

```python
""" 
Question 4: Given the following list of tuples,

a.    sort the list by 1st item in the tuple in ascending order

b.    Sort the list by 2nd item in the tuple in ascending order

c.     Repeat a, b using descending order

listA = [(1,2), (4,3), (2,10), (12, 5), (6, 7), (9,11), (15, 4)]
"""


def sort_tuple_asc(tup, choice):
    n = len(tup)
    for i in range(n):
        for j in range(n-i-1):
            if tup[j][choice] > tup[j + 1][choice]:
                tup[j], tup[j + 1] = tup[j + 1], tup[j]
    return tup

def sort_tuple_desc(tup, choice):
    n = len(tup)
    for i in range(n):
        for j in range(n-i-1):
            if tup[j][choice] < tup[j + 1][choice]:
                tup[j], tup[j + 1] = tup[j + 1], tup[j]
    return tup


listA = [(1,2), (4,3), (2,10), (12, 5), (6, 7), (9,11), (15, 4)]

def list_sort(listA):
  for i in range(2):
    choice = i
    print(f"\nAscending : {choice} element")
    print(sort_tuple_asc(listA, choice))
    print(f"\nDescending : {choice} element")
    print(sort_tuple_desc(listA, choice))

list_sort(listA)
```
